```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os

# 1. Configuration
data_path = '../data/raw/*_prices.csv'
file_list = glob.glob(data_path)
data_frames = []

# 2. Dynamic Loading and Merging
for file in file_list:
    country_name = os.path.basename(file).replace('_prices.csv', '')
    
    df = pd.read_csv(file)
    
    # THE FIX: Drop the repeating 'Position' column completely
    if 'Position' in df.columns:
        df = df.drop(columns=['Position'])
        
    df = df.rename(columns={'Price_EUR_MWh': country_name})
    
    data_frames.append(df)

# 3. Concatenate all countries horizontally (row-by-row)
master_df = pd.concat(data_frames, axis=1)

print(f"Successfully loaded and merged {len(data_frames)} regional datasets.")
print(f"Master Dataset Shape: {master_df.shape[0]} intervals by {master_df.shape[1]} countries.")
display(master_df.head())
```

```python
# 1. Force all data to be numeric (just as a safety net)
master_df = master_df.apply(pd.to_numeric, errors='coerce')

# 2. Generate descriptive statistics
summary_stats = master_df.describe().T

# 3. Sort by the lowest minimum price
summary_stats = summary_stats.sort_values(by='min')

print("--- European Price Statistics (Sorted by Most Extreme Negative Prices) ---")
display(summary_stats[['mean', 'std', 'min', 'max']])
```

```python
import pandas as pd
import glob
import os

# 1. Verify the files actually exist where Jupyter is looking
data_path = '../data/raw/*_prices.csv'
file_list = glob.glob(data_path)

print(f"🔍 Found {len(file_list)} files matching the path.")

if len(file_list) == 0:
    print("🚨 STOP! We found 0 files. Your Jupyter notebook is looking in the wrong folder.")
    print("Try changing data_path to 'data/raw/*_prices.csv' (removing the ../)")
else:
    data_frames = []
    
    # 2. Dynamic Loading, Cleaning, and STRICT Type Casting
    for file in file_list:
        country_name = os.path.basename(file).replace('_prices.csv', '')
        df = pd.read_csv(file)
        
        # Drop the useless position column
        if 'Position' in df.columns:
            df = df.drop(columns=['Position'])
            
        # FORCE the column to be numeric BEFORE we merge it
        df['Price_EUR_MWh'] = pd.to_numeric(df['Price_EUR_MWh'], errors='coerce')
        
        # Rename and append
        df = df.rename(columns={'Price_EUR_MWh': country_name})
        data_frames.append(df)

    # 3. Concatenate all countries
    master_df = pd.concat(data_frames, axis=1)
    print(f"✅ Successfully created Master Table! Shape: {master_df.shape}")
    
    # 4. Generate and Sort Statistics
    summary_stats = master_df.describe().T
    summary_stats = summary_stats.sort_values(by='min')
    
    print("\n--- European Price Statistics ---")
    display(summary_stats[['mean', 'std', 'min', 'max']])
```
